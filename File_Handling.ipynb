```python
# Python File Handling and Executing Linux Commands Through IPython : ----------------------

## Note: If you want to execute any linux commands within the IPython interactive shell, you can use the exclamation mark "!" with the linux command name.

1. !pwd                                          # Finding the working directory in which ipython is currently storing its data.
2. !ls                                           # Finding the existing files within the current working directory.
3. %cd /home/your_username # The Directory       # Changing the current working directory to some other directory.

4. new = open("filename_withextension","w")      # Opening a file in a written mode
5. new.write("Write_Your_Contents")              # Writing some contents in a file
5. new = open("filename_withextension","r")      # Opening a file in a reading mode
6. new.read()                                    # Reading a file
7. new.close()                                   # Closing a file
8. new = open("filename_withextension","a")      # Writing extra contents in a file


# Note 1 : You can override the contents written in your file by opening the same written file in written mode, and writing some contents in it.
# Note 2 : You can add some contents within the file by opening the same written file in append mode, and writing some contents in it which you want to add.
```
